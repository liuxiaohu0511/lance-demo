分词器（Tokenizers）¶
目前，Lance 内置支持 **Jieba** 和 **Lindera** 两种分词器。
但是，它**不自带语言模型**。如果需要使用分词功能，你需要**自行下载语言模型**。
可以通过设置环境变量 **`LANCE_LANGUAGE_MODEL_HOME`** 来指定语言模型的存储位置。
如果未设置该变量，默认路径为：

```
${system data directory}/lance/language_models
```

此外，Lance 还支持**配置用户词典（user dictionaries）**，这使得用户可以方便地扩展自己的词典，而无需重新训练语言模型。

---

### Jieba 的语言模型¶

**下载模型：**

```bash
python -m lance.download jieba
```

语言模型默认存储在：

```
${LANCE_LANGUAGE_MODEL_HOME}/jieba/default
```

**使用模型：**

```python
ds.create_scalar_index("text", "INVERTED", base_tokenizer="jieba/default")
```

**用户词典：**
在当前模型的根目录下创建一个名为 `config.json` 的文件：

```json
{
    "main": "dict.txt",
    "users": ["path/to/user/dict.txt"]
}
```

* `"main"` 字段是可选的，如果不填写，默认值为 `"dict.txt"`。
* `"users"` 是用户词典文件的路径。
  用户词典的格式可参考：
  [https://github.com/messense/jieba-rs/blob/main/src/data/dict.txt](https://github.com/messense/jieba-rs/blob/main/src/data/dict.txt)

### Lindera 的语言模型¶

**下载模型：**

```bash
python -m lance.download lindera -l [ipadic|ko-dic|unidic]
```

请注意，**Lindera 的语言模型需要编译**。
在此之前请先安装 `lindera-cli`。
详细步骤可参考：
[https://github.com/lindera/lindera/tree/main/lindera-cli](https://github.com/lindera/lindera/tree/main/lindera-cli)

语言模型默认存储在：

```
${LANCE_LANGUAGE_MODEL_HOME}/lindera/[ipadic|ko-dic|unidic]
```

**使用模型：**

```python
ds.create_scalar_index("text", "INVERTED", base_tokenizer="lindera/ipadic")
```

**用户词典：**
在模型根目录下创建一个名为 `config.yml` 的文件，
或者通过环境变量 **`LINDERA_CONFIG_PATH`** 指定自定义的 YAML 配置文件路径。
如果两者都存在，则会优先使用模型根目录下的 `config.yml`。

更详细的配置方法请参阅 Lindera 官方文档：
[https://github.com/lindera/lindera/](https://github.com/lindera/lindera/)

示例配置：

```yaml
segmenter:
    mode: "normal"
    dictionary:
        # 注意：在 Lance 中，`kind` 字段不受支持。
        # 需要使用 `path` 字段指定模型路径。
        path: /path/to/lindera/ipadic/main
```
